```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset
df = pd.read_csv('/content/compact.txt')

# Display success message
print("COVID-19 Dataset Loaded Successfully!")

# Display first 5 rows
display(df.head())
```

```python
# Check dataset size
print("Dataset Shape:", df.shape)

# Check column names
print("\nColumn Names:")
print(df.columns.tolist())

# Check dataset information
print("\nDataset Information:")
df.info()
```

```python
required_columns = [
    'location',
    'date',
    'total_cases',
    'total_deaths',
    'total_vaccinations'
]

for col in required_columns:
    print(col, ":", col in df.columns)
```

```python
print("Missing Values in Each Column:")
print(df.isnull().sum())
```

```python
print("Duplicate Rows:", df.duplicated().sum())
```

```python
df['date'] = pd.to_datetime(df['date'], errors='coerce')
```

```python
numeric_columns = [
    'total_cases',
    'total_deaths',
    'total_vaccinations'
]

for col in numeric_columns:
    df[col] = pd.to_numeric(df[col], errors='coerce')
```

```python
df = df.drop_duplicates()
```

```python
df = df.dropna(subset=['location', 'date'])
```

```python
df = df.sort_values(by='date')

print("Data Cleaning Completed!")
```

```python
# Convert date into datetime
df['date'] = pd.to_datetime(df['date'])

# Get latest record for each country
latest_data = (
    df.sort_values('date')
      .groupby('country')
      .tail(1)
)

print(latest_data.head())
```

```python
top_5 = (
    latest_data
    .sort_values('total_cases', ascending=False)
    .head(5)
)

print("Top 5 Countries by COVID-19 Cases:")
print(top_5[['country', 'total_cases']])
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(12, 6))

sns.barplot(
    data=top_5,
    x='country',
    y='total_cases'
)

plt.title("Top 5 Countries by Total COVID-19 Cases")
plt.xlabel("Country")
plt.ylabel("Total COVID-19 Cases")

plt.xticks(rotation=30)
plt.tight_layout()

plt.show()
```

```python
heatmap_data = latest_data[
    ['total_cases', 'total_deaths', 'total_vaccinations']
].copy()

heatmap_data = heatmap_data.fillna(0)
```

```python
correlation = heatmap_data.corr()

print(correlation)
```

```python
plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation,
    annot=True,
    cmap='coolwarm',
    fmt='.2f',
    linewidths=0.5
)

plt.title("COVID-19 Data Correlation Heatmap")

plt.tight_layout()
plt.show()
```

```python
scatter_data = latest_data[
    ['country', 'total_cases', 'total_deaths']
].dropna()
```

```python
plt.figure(figsize=(10, 7))

sns.scatterplot(
    data=scatter_data,
    x='total_cases',
    y='total_deaths',
    s=60
)

plt.title("COVID-19 Cases vs Deaths")
plt.xlabel("Total COVID-19 Cases")
plt.ylabel("Total COVID-19 Deaths")

plt.xscale('log')
plt.yscale('log')

plt.grid(True, alpha=0.3)
plt.tight_layout()

plt.show()
```

```python
import os

os.makedirs("output", exist_ok=True)

# Save line chart
plt.figure(figsize=(12, 6))

for country in top_5['country']:
    country_data = df[df['country'] == country]
    plt.plot(
        country_data['date'],
        country_data['total_cases'],
        label=country
    )

plt.title("COVID-19 Cases Trend Over Time")
plt.xlabel("Date")
plt.ylabel("Total Cases")
plt.legend()
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig("output/covid_trend.png", dpi=300)
plt.show()

print("Line chart saved!")
```
